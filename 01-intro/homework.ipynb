```python
df = pd.read_csv("car_fuel_efficiency.csv")
```

```python
records_count = len(df)
```

```python
print(f"2. Records count: {records_count}")
```

```python
if 'fuel_type' in df.columns:
    fuel_types_count = df['fuel_type'].nunique()
    print(f"3. Unique fuel types: {fuel_types_count}")
else:
    print("3. 'fuel_type' column not found. Available columns:", list(df.columns))
```

```python
total_missing = df.isnull().sum()
print(f"4. Total missing values: {total_missing}")
```

```python
fe_col = [col for col in df.columns if 'fuel' in col.lower() or 'mpg' in col.lower()]
if fe_col:
    max_fe = df[fe_col[-1]].max()
    print(f"5. Max fuel efficiency: {max_fe}")
else:
    print("5. Fuel efficiency column not automatically detected.")
```

```python
hp_col = [df.columns[-4]]
median_before = df[hp_col[0]].median()
median_after = df.dropna(subset=[hp_col[0]])[hp_col[0]].median()
print(f"6. Median horsepower (Raw): {median_before}")
print(f"   Median horsepower (After dropping NaNs): {median_after}")
if median_after > median_before:
    print("   -> It increased")
elif median_after < median_before:
    print("   -> It decreased")
else:
    print("   -> No change")
    
```

```python
weight_col = [col for col in df.columns if 'weight' in col.lower()]
if weight_col:
    sum_weights = df[weight_col[0]].sum()
    print(f"7. Sum of weights: {sum_weights}")
else:
    print("7. Weight column not automatically detected.")
```
